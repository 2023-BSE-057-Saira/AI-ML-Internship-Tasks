Iris Dataset Analysis and Visualization

**Objective**

Explore and visualize the Iris dataset using Pandas, Matplotlib, and Seaborn to understand data distributions, feature relationships, and potential outliers.

**1. Import Required Libraries**

**Purpose**

Import all necessary libraries for data loading, analysis, and visualization.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

**2. Load the Dataset**

**Purpose**

Load the Iris dataset into a Pandas DataFrame.

```python
iris = sns.load_dataset('iris')

print("Dataset Loaded Successfully!")
```

**3. Display Dataset Shape**

**Purpose**

Check the number of rows and columns in the dataset.

```python
print("Dataset Shape:")
print(iris.shape)
```

**4. Display Column Names**

**Purpose**

View all feature names available in the dataset.

```python
print("Column Names:")
print(iris.columns)
```

**5. Display First Five Rows**

**Purpose**

Inspect the first few records of the dataset.

```python
iris.head()
```

**6. Dataset Information**

**Purpose**

Understand data types, non-null values, and overall structure.

```python
iris.info()
```

**7. Summary Statistics**

**Purpose**

Generate descriptive statistics for numerical features.

```python
iris.describe()
```

**8. Check Missing Values**

**Purpose**

Verify whether the dataset contains any missing values.

```python
print("Missing Values:")
print(iris.isnull().sum())
```

**9. Scatter Plot**

**Purpose**

Visualize the relationship between Sepal Length and Petal Length.

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=iris,
    x='sepal_length',
    y='petal_length',
    hue='species'
)

plt.title("Sepal Length vs Petal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")

plt.show()
```

**10. Histogram Visualization**

**Purpose**

Observe the distribution of all numerical features.

```python
iris.hist(figsize=(12,8))

plt.suptitle("Distribution of Iris Features")

plt.show()
```

**11. Individual Histogram Using Seaborn**

**Purpose**

Analyze the distribution of Sepal Length.

```python
plt.figure(figsize=(8,6))

sns.histplot(
    iris['sepal_length'],
    bins=10,
    kde=True
)

plt.title("Distribution of Sepal Length")

plt.show()
```

**12. Box Plot for Outlier Detection**

**Purpose**

Identify potential outliers in numerical features.

```python
plt.figure(figsize=(10,6))

sns.boxplot(data=iris)

plt.title("Box Plot of Iris Features")

plt.show()
```

**13. Pair Plot Visualization**

**Purpose**

Visualize relationships among all features simultaneously.

```python
sns.pairplot(
    iris,
    hue='species'
)

plt.show()
```

**14. Correlation Matrix**

**Purpose**

Measure relationships between numerical features.

```python
correlation_matrix = iris.drop('species', axis=1).corr()

print(correlation_matrix)
```

**15. Correlation Heatmap**

**Purpose**

Visualize feature correlations using color intensity.

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.show()
```
